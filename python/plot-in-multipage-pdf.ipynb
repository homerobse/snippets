```python
from matplotlib.backends.backend_pdf import PdfPages
pp = PdfPages('plot-in-multipage-pdf.pdf')
import matplotlib.pyplot as plt

# add title page
firstPage = plt.figure(figsize=(11.69,8.27))
# firstPage.clf()
txt = 'This is the title page'
firstPage.text(0.5,0.5,txt, transform=firstPage.transFigure, size=24, ha="center")
pp.savefig()

# add first plot
plt.figure()
plt.plot([1,2,20], '.--')
plt.text(0.05,0.95,"normal plot", transform=gcf().transFigure, size=20)
pp.savefig()

# add second plot
plt.figure()
plt.text(0.05,0.95,"pcolor plot", transform=gcf().transFigure, size=20)
plt.pcolor(array([1,2,3]), array([5,6,7]), array([[1,10],[5,2]]))
pp.savefig()
pp.close()
```
