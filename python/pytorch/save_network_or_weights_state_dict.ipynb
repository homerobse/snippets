```python
import torch
```

```python
class Net(torch.nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.l1 = torch.nn.Linear(3,4)
    def forward(self, x):
        return self.l1(x)

netw = Net()
```

```python
netw.state_dict()
```

```python
torch.save(netw.state_dict(), "saved_model_state_dict.pth")
torch.save(netw, "saved_model_obj.pth")
```

```python
loaded_state_dict = torch.load("saved_model_state_dict.pth")
loaded_model_obj = torch.load("saved_model_obj.pth")
loaded_state_dict, loaded_model_obj
```

```python
model_from_state_dict = Net()
model_from_state_dict.state_dict()
```

```python
model_from_state_dict.load_state_dict(loaded_state_dict)
```
