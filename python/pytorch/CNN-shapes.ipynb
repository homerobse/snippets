```python
import torch
```

```python
class Net(torch.nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.l1 = torch.nn.Conv2d(2,4,3, padding=1)
    def forward(self, x):
        return self.l1(x)

netw = Net()
```

```python
x = torch.rand((1,2, 15,15))
y = netw(x)
```

```python
x.shape, y.shape
```

* Input size can vary
* Padding 1 makes a kernel_size=3 convolution create a feature map of the same size of the input (if the stride is 1)
